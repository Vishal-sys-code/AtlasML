To find the HTML Tags in the given string by using the regex we can do that by this pattern matching that is: `<.*?>`

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)
```

# **3. Remove URLs**

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

# **4. Remove Punctuation**

All the punctuation marks: `'!"#$%&\'()*+,-./:;<=>?@[\\]^{|}~'`

```python
import string,time
exclude = string.punctuation
print(exclude)
```

```python
def remove_punctuations(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
text = 'string, with, punctuation!'
start = time.time()
print(f"After Removing Punctuations: {remove_punctuations(text)}")
total_time1 = time.time() - start
print(f"The total time taken to run this: {total_time1} ms")
```

```python
def remove_Punctuations_optimal(text):
    return text.translate(str.maketrans('','',exclude))
```

```python
text = 'string, with, punctuation!'
start = time.time()
print(f"After Removing Punctuations: {remove_Punctuations_optimal(text)}")
total_time2 = time.time() - start
print(f"The total time taken to run this: {total_time2} ms")
```

```python
total_time1 / total_time2
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in internet_slang:
            new_text.append(internet_slang[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

# **7. Removing Stop Words**

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append(' ')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return "".join(x)
```

```python
remove_stopwords('probably my all-time favorite cartoons, doraemon, shinchan, buru-buri zaemon, action kamen, quantum robot')
```

# **8. Handling Emojies**

```python
import re
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags (iOS)
        "\U00002700-\U000027BF"  # Dingbats
        "\U0001F900-\U0001F9FF"  # Supplemental Symbols and Pictographs
        "\U00002600-\U000026FF"  # Misc symbols
        "\U0001FA70-\U0001FAFF"  # Symbols and Pictographs Extended-A
        "\U000025A0-\U000025FF"  # Geometric Shapes
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
```

```python
remove_emojis("Our team won the match! 🏆⚽🇧🇷")
```

# **10. Stemming and Lemmatization**

Stemming is the process of reducing inflection in qords to their root forms such as mapping a group of words to the same stem even if the stem itself is not a valid word in the Language

Lemmatization, unlike Stemming, reduces the inflected words properly ensuring that the root word belong to the language. In Lemmatization root word is called Lemma. A lemma is the canonical form, dictionary form, or citation form of a set of words.

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = 'walk walks walking walked'
stem_words(sample)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = "?;!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)
sentence_words 
print("{0:20}{1:20}".format("Word", "Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word, wordnet_lemmatizer.lemmatize(word)))
```
