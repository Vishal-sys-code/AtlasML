```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        'text': ['people watch space',
                 'space watch space',
                 'people write comment',
                 'space write people'
                ],
        'output': [1,1,0,0]
    }
)

df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
count_vector = CountVectorizer()
```

```python
bow = count_vector.fit_transform(df['text'])
```

```python
#Vocab
print(count_vector.vocabulary_)
```

## **For unigrams**

```python
count_vector = CountVectorizer(ngram_range = (1, 1))
bow = count_vector.fit_transform(df['text'])
print(count_vector.vocabulary_)
```

## **For Bigrams**

```python
count_vector = CountVectorizer(ngram_range = (2, 2))
bow = count_vector.fit_transform(df['text'])
print(count_vector.vocabulary_)
```

## **For Trigrams**

```python
count_vector = CountVectorizer(ngram_range = (2, 2))
bow = count_vector.fit_transform(df['text'])
print(count_vector.vocabulary_)
```

## **Some special cases**

```python
count_vector = CountVectorizer(ngram_range = (1, 2))
bow = count_vector.fit_transform(df['text'])
print(count_vector.vocabulary_)
```
