```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        'text': ['people watch space',
                 'space watch space',
                 'people write comment',
                 'space write people'
                ],
        'output': [1,1,0,0]
    }
)

df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
tfidf.fit_transform(df['text']).toarray()
```

```python
print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```
