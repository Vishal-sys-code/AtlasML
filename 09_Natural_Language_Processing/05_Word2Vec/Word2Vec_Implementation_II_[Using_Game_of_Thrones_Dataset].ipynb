```python
import numpy as np
import pandas as pd
import gensim
import os
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
story = []
for filename in os.listdir(path):
  f = open(os.path.join(path, filename), encoding='latin-1')
  corpus = f.read()
  raw_sent = sent_tokenize(corpus)
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

Path mei se har ek file utha raha hai aur read kar raha hai and corpus mei store kar raha hai.
Content ko corpus bolte hai. Phir humlog sentences mei divide krenge, by using tokenize.
Then we will preprocess it and then we will append to the story_list

```python
model = gensim.models.Word2Vec(
    window = 10,
    min_count = 2,
    workers = 4
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples = model.corpus_count, epochs = model.epochs)
```
